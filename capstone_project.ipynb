```python
# ==========================================
# SECTION 1: DATA INSPECTION
# ==========================================

# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Ignore warning messages
warnings.filterwarnings("ignore")


# ------------------------------------------
# Load Dataset
# ------------------------------------------

# Read the customer shopping dataset
customers = pd.read_csv("shopping_trends.csv")

# Create a working copy of the dataset
clean = customers.copy()


# ------------------------------------------
# Display First Five Rows
# ------------------------------------------

print("FIRST FIVE ROWS")
display(clean.head())


# ------------------------------------------
# Display Last Five Rows
# ------------------------------------------

print("\nLAST FIVE ROWS")
display(clean.tail())


# ------------------------------------------
# Check Dataset Dimensions
# ------------------------------------------

print("\nDATASET DIMENSIONS")
print(f"Rows: {clean.shape[0]}")
print(f"Columns: {clean.shape[1]}")


# ------------------------------------------
# Display Column Names
# ------------------------------------------

print("\nCOLUMN NAMES")
print(clean.columns)


# ------------------------------------------
# Display Data Types
# ------------------------------------------

print("\nDATA TYPES")
display(clean.dtypes)


# ------------------------------------------
# Check Missing Values
# ------------------------------------------

print("\nMISSING VALUES")
display(clean.isnull().sum())


# ------------------------------------------
# Check Duplicate Records
# ------------------------------------------

print("\nDUPLICATE RECORDS")
print(clean.duplicated().sum())


# ------------------------------------------
# Generate Descriptive Statistics
# ------------------------------------------

print("\nDESCRIPTIVE STATISTICS")
display(clean.describe(include="all"))
```

```python
# ==========================================
# SECTION 2: DATA CLEANING
# ==========================================


# Create a working copy of the dataset
clean = customers.copy()


# ------------------------------------------
# 1. Remove Duplicate Records
# ------------------------------------------

print("DUPLICATE RECORDS BEFORE CLEANING")
print(clean.duplicated().sum())

# Remove duplicate records
clean = clean.drop_duplicates()

print("\nDUPLICATE RECORDS AFTER CLEANING")
print(clean.duplicated().sum())


# ------------------------------------------
# 2. Handle Missing Values
# ------------------------------------------

print("\nMISSING VALUES BEFORE CLEANING")
display(clean.isnull().sum())

# Remove rows with missing values
clean = clean.dropna()

print("\nMISSING VALUES AFTER CLEANING")
display(clean.isnull().sum())


# ------------------------------------------
# 3. Standardize Gender
# ------------------------------------------

# Remove extra spaces
# Convert gender values to uppercase

clean["Gender"] = (
    clean["Gender"]
    .str.strip()
    .str.upper()
)


# ------------------------------------------
# 4. Standardize Customer Locations
# ------------------------------------------

# Remove unnecessary spaces

clean["Location"] = (
    clean["Location"]
    .str.strip()
)


# ------------------------------------------
# 5. Convert Purchase Amount into Numeric Format
# ------------------------------------------

clean["Purchase Amount (USD)"] = pd.to_numeric(
    clean["Purchase Amount (USD)"],
    errors="coerce"
)


# ------------------------------------------
# 6. Remove Impossible Ages
# ------------------------------------------

# Keep only realistic customer ages

clean = clean[
    (clean["Age"] >= 18) &
    (clean["Age"] <= 100)
]


# ------------------------------------------
# 7. Standardize Review Ratings
# ------------------------------------------

clean["Review Rating"] = pd.to_numeric(
    clean["Review Rating"],
    errors="coerce"
)


# ------------------------------------------
# 8. Final Data Cleaning Check
# ------------------------------------------

print("\nFINAL DATASET INFORMATION")
clean.info()

print("\nFIRST FIVE ROWS OF CLEANED DATASET")
display(clean.head())
```

```python
# ==========================================
# SECTION 3: EXPLORATORY DATA ANALYSIS (EDA)
# ==========================================


# Improve chart appearance
plt.style.use("ggplot")


# ------------------------------------------
# 1. Customer Age Distribution
# ------------------------------------------

plt.figure(figsize=(8,5))
sns.histplot(clean["Age"], bins=10, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.show()



# ------------------------------------------
# 2. Customer Satisfaction Ratings
# ------------------------------------------

plt.figure(figsize=(8,5))
sns.countplot(data=clean, x="Review Rating")

plt.title("Customer Satisfaction Ratings")
plt.xlabel("Review Rating")
plt.ylabel("Number of Customers")

plt.show()



# ------------------------------------------
# 3. Average Spending by Gender
# ------------------------------------------

average_spending = clean.groupby("Gender")["Purchase Amount (USD)"].mean()

plt.figure(figsize=(6,5))
average_spending.plot(kind="bar")

plt.title("Average Spending by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Purchase Amount (USD)")

plt.show()



# ------------------------------------------
# 4. Income (Purchase Amount) vs Age
# ------------------------------------------

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=clean,
    x="Age",
    y="Purchase Amount (USD)"
)

plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")

plt.show()



# ------------------------------------------
# 5. Customer Distribution by Location
# ------------------------------------------

plt.figure(figsize=(12,5))
sns.countplot(
    data=clean,
    x="Location",
    order=clean["Location"].value_counts().index
)

plt.title("Customers by Location")
plt.xticks(rotation=45)
plt.xlabel("Location")
plt.ylabel("Number of Customers")

plt.show()



# ------------------------------------------
# 6. Preferred Payment Method
# ------------------------------------------

plt.figure(figsize=(8,5))
sns.countplot(data=clean, x="Payment Method")

plt.title("Preferred Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")

plt.show()



# ------------------------------------------
# 7. Correlation Heatmap
# ------------------------------------------

plt.figure(figsize=(8,6))

sns.heatmap(
    clean.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()



# ------------------------------------------
# 8. Subscription Status
# ------------------------------------------

plt.figure(figsize=(6,5))
sns.countplot(data=clean, x="Subscription Status")

plt.title("Subscription Status")
plt.xlabel("Subscription")
plt.ylabel("Number of Customers")

plt.show()



# ------------------------------------------
# 9. Most Purchased Item Category
# ------------------------------------------

plt.figure(figsize=(10,5))
sns.countplot(
    data=clean,
    x="Category",
    order=clean["Category"].value_counts().index
)

plt.title("Most Purchased Product Categories")
plt.xlabel("Category")
plt.ylabel("Number of Purchases")

plt.show()
```
